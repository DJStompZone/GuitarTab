 # Token Visualization



 This notebook visualizes parsed tokens as human-readable scores:

 1. **Six-string tablature notation** - shows fret numbers on each string with bars

 2. **Note name notation** - shows note names with octave numbers and rhythm



 Both formats display bars and rhythm correctly.

```python
import random
import sys
from pathlib import Path

# Add parent directory to path
sys.path.insert(0, str(Path.cwd().parent))

from src.dadagp_parser import parse_dadagp_file_to_events
from src.visualization import render_as_notes, render_as_tablature

print("✓ Imports successful")
```

 ## 1. Find Sample Files

```python
# Find token files
dadagp_dir = Path("../DadaGP-v1.1")
token_files = list(dadagp_dir.rglob("*.tokens.txt"))

print(f"Found {len(token_files):,} token files")

if token_files:
    # Select a few files for visualization
    sample_files = random.sample(token_files, min(3, len(token_files)))
    print(f"\nSelected {len(sample_files)} files:")
    for i, f in enumerate(sample_files, 1):
        print(f"  {i}. {f.relative_to(dadagp_dir)}")
else:
    print("No token files found")
    sample_files = []
```

 ## 2. Parse First Sample File

```python
if sample_files:
    sample_file = sample_files[0]
    print(f"Parsing: {sample_file.name}")
    print("=" * 60)

    # Parse into events
    input_events, output_events = parse_dadagp_file_to_events(str(sample_file))

    print(f"Input events:  {len(input_events):,} tokens")
    print(f"Output events: {len(output_events):,} tokens")
    print("\n✓ Parsed successfully")
else:
    print("No files to parse")
    input_events = None
    output_events = None
```

 ## 3. Visualization 1: Six-String Tablature



 Shows fret numbers on each guitar string with proper bar lines and rhythm.

```python
if output_events:
    print("\n" + "=" * 80)
    print("TABLATURE NOTATION")
    print("=" * 80 + "\n")

    tab_viz = render_as_tablature(output_events, max_bars=4, chars_per_beat=8)
    print(tab_viz)
else:
    print("No output events to visualize")
```

 ## 4. Visualization 2: Note Names with Rhythm



 Shows note names (e.g., C4, E3) with duration and timing.

```python
if input_events:
    print("\n" + "=" * 80)
    print("NOTE NOTATION")
    print("=" * 80 + "\n")

    note_viz = render_as_notes(input_events, max_bars=4, chars_per_beat=8)
    print(note_viz)
else:
    print("No input events to visualize")
```

 ## 5. Side-by-Side Comparison



 Show both visualizations for the first 2 bars only.

```python
if input_events and output_events:
    print("\n" + "=" * 80)
    print("SIDE-BY-SIDE COMPARISON (First 2 Bars)")
    print("=" * 80 + "\n")

    print("TABLATURE:")
    print("-" * 80)
    print(render_as_tablature(output_events, max_bars=2, chars_per_beat=8))

    print("\n" + "-" * 80)
    print("\nNOTES:")
    print("-" * 80)
    print(render_as_notes(input_events, max_bars=2, chars_per_beat=8))
```

 ## 6. Interactive Visualization Function



 Easily visualize any file from the dataset.

```python
def visualize_file(file_index=0, max_bars=3, chars_per_beat=8):
    """
    Visualize a specific file from the sample files.

    Args:
        file_index: Index in sample_files list (0-based)
        max_bars: Number of bars to display
        chars_per_beat: Characters per beat for spacing
    """
    if not sample_files:
        print("No sample files available")
        return

    if file_index >= len(sample_files):
        print(f"Index {file_index} out of range (max: {len(sample_files) - 1})")
        return

    file = sample_files[file_index]
    print(f"File: {file.name}")
    print("=" * 80)

    # Parse
    inp, out = parse_dadagp_file_to_events(str(file))
    print(f"Events: {len(inp):,} input, {len(out):,} output\n")

    # Visualize tablature
    print("\n" + "=" * 80)
    print("TABLATURE VIEW")
    print("=" * 80)
    print(render_as_tablature(out, max_bars=max_bars, chars_per_beat=chars_per_beat))

    # Visualize notes
    print("\n\n" + "=" * 80)
    print("NOTE VIEW")
    print("=" * 80)
    print(render_as_notes(inp, max_bars=max_bars, chars_per_beat=chars_per_beat))


print("✓ Function defined")
print("\nUsage: visualize_file(file_index=0, max_bars=3, chars_per_beat=8)")
```

 ## 7. Try Different Files



 Use the function to explore different files.

```python
# Visualize first file (index 0)
visualize_file(file_index=0, max_bars=3, chars_per_beat=8)
```

```python
# Try second file if available
if len(sample_files) > 1:
    visualize_file(file_index=1, max_bars=2, chars_per_beat=10)
```

 ## 8. Customize Visualization Parameters



 Experiment with different visualization settings.

```python
if output_events:
    print("Compact view (more bars, less spacing):")
    print("=" * 80)
    print(render_as_tablature(output_events, max_bars=6, chars_per_beat=6))

    print("\n\nExpanded view (fewer bars, more spacing):")
    print("=" * 80)
    print(render_as_tablature(output_events, max_bars=2, chars_per_beat=12))
```

 ## 9. Batch Visualization



 Quick preview of multiple files.

```python
if sample_files:
    print("Quick preview of all sample files (first bar only):")
    print("=" * 80)

    for i, f in enumerate(sample_files):
        try:
            inp, out = parse_dadagp_file_to_events(str(f))
            print(f"\n[{i + 1}] {f.name}")
            print("-" * 80)
            print(render_as_tablature(out, max_bars=1, chars_per_beat=8))
        except Exception as e:
            print(f"\n[{i + 1}] {f.name} - Error: {e}")
```

 ## 10. Export Visualization to Text File



 Save a visualization for later reference.

```python
def export_visualization(
    file_index=0, output_path="../visualization_output.txt", max_bars=10
):
    """
    Export visualization of a file to a text file.

    Args:
        file_index: Index in sample_files list
        output_path: Path to save the output
        max_bars: Number of bars to export
    """
    if not sample_files or file_index >= len(sample_files):
        print("Invalid file index")
        return

    file = sample_files[file_index]
    inp, out = parse_dadagp_file_to_events(str(file))

    with open(output_path, "w") as f:
        f.write(f"File: {file.name}\n")
        f.write(f"Events: {len(inp):,} input, {len(out):,} output\n")
        f.write("=" * 80 + "\n\n")

        f.write("TABLATURE NOTATION\n")
        f.write("=" * 80 + "\n")
        f.write(render_as_tablature(out, max_bars=max_bars, chars_per_beat=8))
        f.write("\n\n")

        f.write("NOTE NOTATION\n")
        f.write("=" * 80 + "\n")
        f.write(render_as_notes(inp, max_bars=max_bars, chars_per_beat=8))

    print(f"✓ Visualization exported to: {output_path}")


# Example: export first file
if sample_files:
    export_visualization(
        file_index=0, output_path="../sample_visualization.txt", max_bars=5
    )
    print("\nYou can view the file: cat ../sample_visualization.txt")
```
